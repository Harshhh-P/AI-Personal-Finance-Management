```python
# Create a separate working copy
data = df.copy()

print("Original df shape:", df.shape)
print("Working data shape:", data.shape)
```

```python
# Convert Date column to datetime
data["Date"] = pd.to_datetime(data["Date"])

print("Date data type:", data["Date"].dtype)
print("Missing dates:", data["Date"].isna().sum())
```

```python
# Create a separate dataset containing only expenses
expenses = data[data["Type"] == "Expense"].copy()

print("Expense dataset shape:", expenses.shape)
print("\nTransaction types:")
print(expenses["Type"].value_counts())
```

```python
# Create time-based features
expenses["Year"] = expenses["Date"].dt.year
expenses["Month"] = expenses["Date"].dt.month
expenses["Month_Name"] = expenses["Date"].dt.month_name()
expenses["Year_Month"] = expenses["Date"].dt.to_period("M")

print(expenses[["Date", "Year", "Month", "Month_Name", "Year_Month"]].head())
```

```python
print("Number of months:", expenses["Year_Month"].nunique())

print("\nFirst month:", expenses["Year_Month"].min())
print("Last month:", expenses["Year_Month"].max())
```

```python
# Monthly expense aggregation

monthly_expenses = expenses.groupby("Year_Month").agg(
    Total_Expense=("Amount", "sum"),
    Transaction_Count=("Amount", "count"),
    Average_Expense=("Amount", "mean"),
    Max_Expense=("Amount", "max"),
    Min_Expense=("Amount", "min")
).reset_index()

print("Monthly dataset shape:", monthly_expenses.shape)

print("\nFirst 5 months:")
print(monthly_expenses.head())
```

```python
print("\nLast 5 months:")
print(monthly_expenses.tail())

print("\nMissing values:")
print(monthly_expenses.isnull().sum())

print("\nData types:")
print(monthly_expenses.dtypes)
```

```python
# Create numerical time features

monthly_expenses["Year"] = monthly_expenses["Year_Month"].dt.year
monthly_expenses["Month"] = monthly_expenses["Year_Month"].dt.month

print(monthly_expenses.head())
print("\nData types:")
print(monthly_expenses.dtypes)
```

```python
# Create lag features using only previous months

monthly_expenses["Previous_Month_Expense"] = (
    monthly_expenses["Total_Expense"].shift(1)
)

monthly_expenses["Expense_2_Months_Ago"] = (
    monthly_expenses["Total_Expense"].shift(2)
)

monthly_expenses["Expense_3_Months_Ago"] = (
    monthly_expenses["Total_Expense"].shift(3)
)

monthly_expenses["Previous_Transaction_Count"] = (
    monthly_expenses["Transaction_Count"].shift(1)
)

# 3-month rolling average using previous months only
monthly_expenses["Three_Month_Avg_Expense"] = (
    monthly_expenses["Total_Expense"].shift(1).rolling(window=3).mean()
)

print(monthly_expenses.head(7))
```

```python
# Create the dataset for machine learning
# We need 3 months of previous data for our lag features

model_data = monthly_expenses.dropna().copy()

print("Original monthly rows:", len(monthly_expenses))
print("Rows available for ML:", len(model_data))

print("\nMissing values:")
print(model_data.isnull().sum())
```

```python
print(model_data.head())

print("\nShape:", model_data.shape)
```

```python
# Define features and target

features = [
    "Year",
    "Month",
    "Previous_Month_Expense",
    "Expense_2_Months_Ago",
    "Expense_3_Months_Ago",
    "Previous_Transaction_Count",
    "Three_Month_Avg_Expense"
]

X = model_data[features]
y = model_data["Total_Expense"]

print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)

print("\nFeatures:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
# Time-based train-test split

split_index = int(len(model_data) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training data:")
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("\nTesting data:")
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

print("\nTraining period:")
print(model_data["Year_Month"].iloc[0], "to", model_data["Year_Month"].iloc[split_index - 1])

print("\nTesting period:")
print(model_data["Year_Month"].iloc[split_index], "to", model_data["Year_Month"].iloc[-1])
```
