```python
processed_path = "/content/drive/MyDrive/AI-Personal-Finance-Management/ml/datasets/processed/monthly_expenses_ml_ready.csv"

model_data = pd.read_csv(processed_path)

print("Processed dataset loaded successfully!")
print("Shape:", model_data.shape)
print("\nColumns:")
print(model_data.columns.tolist())
```

```python
features = [
    "Year",
    "Month",
    "Previous_Month_Expense",
    "Expense_2_Months_Ago",
    "Expense_3_Months_Ago",
    "Previous_Transaction_Count",
    "Three_Month_Avg_Expense"
]

target = "Total_Expense"

X = model_data[features]
y = model_data[target]

print("Features defined successfully!")
print("\nFeatures:")
print(features)

print("\nTarget:")
print(target)

print("\nX shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor

final_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)

final_model.fit(X, y)

print("Final Random Forest model trained successfully!")
print("Number of trees:", final_model.n_estimators)
print("Training samples:", len(X))
```

```python
import os

models_path = "/content/drive/MyDrive/AI-Personal-Finance-Management/ml/models"

os.makedirs(models_path, exist_ok=True)

print("Models folder ready!")
print("Path:", models_path)
```

```python
model_file = os.path.join(
    models_path,
    "random_forest_expense_model.pkl"
)

joblib.dump(
    final_model,
    model_file
)

print("Model saved successfully!")
print("Model path:", model_file)
```

```python
loaded_model = joblib.load(model_file)

print("Saved model loaded successfully!")
print("Model type:", type(loaded_model).__name__)
print("Number of trees:", loaded_model.n_estimators)
```

```python
import joblib

model_file = "/content/drive/MyDrive/AI-Personal-Finance-Management/ml/models/random_forest_expense_model.pkl"
features_file = "/content/drive/MyDrive/AI-Personal-Finance-Management/ml/models/model_features.pkl"
metadata_file = "/content/drive/MyDrive/AI-Personal-Finance-Management/ml/models/model_metadata.pkl"

loaded_model = joblib.load(model_file)
loaded_features = joblib.load(features_file)
loaded_metadata = joblib.load(metadata_file)

print("All deployment files loaded successfully!")

print("\nModel:")
print(type(loaded_model).__name__)

print("\nFeatures:")
print(loaded_features)

print("\nMetadata:")
print(loaded_metadata)
```

```python
sample_input = pd.DataFrame([{
    "Year": 2025,
    "Month": 1,
    "Previous_Month_Expense": 10938.92,
    "Expense_2_Months_Ago": 23565.96,
    "Expense_3_Months_Ago": 22984.21,
    "Previous_Transaction_Count": 20,
    "Three_Month_Avg_Expense": 19163.03
}])

sample_prediction = loaded_model.predict(sample_input)

print("Sample prediction successful!")
print("Predicted monthly expense: ₹", round(sample_prediction[0], 2))
```

```python
def predict_monthly_expense(
    year,
    month,
    previous_month_expense,
    expense_2_months_ago,
    expense_3_months_ago,
    previous_transaction_count,
    three_month_avg_expense
):
    input_data = pd.DataFrame([{
        "Year": year,
        "Month": month,
        "Previous_Month_Expense": previous_month_expense,
        "Expense_2_Months_Ago": expense_2_months_ago,
        "Expense_3_Months_Ago": expense_3_months_ago,
        "Previous_Transaction_Count": previous_transaction_count,
        "Three_Month_Avg_Expense": three_month_avg_expense
    }])

    prediction = loaded_model.predict(input_data)

    return round(float(prediction[0]), 2)


print("Prediction function created successfully!")
```

```python
predicted_expense = predict_monthly_expense(
    year=2025,
    month=1,
    previous_month_expense=10938.92,
    expense_2_months_ago=23565.96,
    expense_3_months_ago=22984.21,
    previous_transaction_count=20,
    three_month_avg_expense=19163.03
)

print("Prediction function tested successfully!")
print("Predicted monthly expense: ₹", predicted_expense)
```
